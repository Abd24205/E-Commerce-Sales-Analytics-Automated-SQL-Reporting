```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from pathlib import Path

DATA_PATH = Path("../data/raw")

print("Files found:")
for file in DATA_PATH.glob("*.csv"):
    print(file.name)
```

```python
customers = pd.read_csv(DATA_PATH / "olist_customers_dataset.csv")
orders = pd.read_csv(DATA_PATH / "olist_orders_dataset.csv")
order_items = pd.read_csv(DATA_PATH / "olist_order_items_dataset.csv")
products = pd.read_csv(DATA_PATH / "olist_products_dataset.csv")
payments = pd.read_csv(DATA_PATH / "olist_order_payments_dataset.csv")
reviews = pd.read_csv(DATA_PATH / "olist_order_reviews_dataset.csv")
sellers = pd.read_csv(DATA_PATH / "olist_sellers_dataset.csv")
```

```python
datasets = {
    "customers": customers,
    "orders": orders,
    "order_items": order_items,
    "products": products,
    "payments": payments,
    "reviews": reviews,
    "sellers": sellers
}

for name, df in datasets.items():
    print(f"{name}: {df.shape[0]:,} rows × {df.shape[1]} columns")
```

```python
orders['order_purchase_timestamp'] = pd.to_datetime(
    orders['order_purchase_timestamp']
)

print("Start date:", orders['order_purchase_timestamp'].min())
print("End date:", orders['order_purchase_timestamp'].max())
```

```python
customers_clean = customers.copy()
orders_clean = orders.copy()
order_items_clean = order_items.copy()
products_clean = products.copy()
payments_clean = payments.copy()
reviews_clean = reviews.copy()
sellers_clean = sellers.copy()
```

```python
date_columns = {
    "orders": [
        "order_purchase_timestamp",
        "order_approved_at",
        "order_delivered_carrier_date",
        "order_delivered_customer_date",
        "order_estimated_delivery_date"
    ],
    "order_items": [
        "shipping_limit_date"
    ],
    "reviews": [
        "review_creation_date",
        "review_answer_timestamp"
    ]
}

for col in date_columns["orders"]:
    orders_clean[col] = pd.to_datetime(orders_clean[col])

for col in date_columns["order_items"]:
    order_items_clean[col] = pd.to_datetime(order_items_clean[col])

for col in date_columns["reviews"]:
    reviews_clean[col] = pd.to_datetime(reviews_clean[col])
```

Delivery time

```python
orders_clean["delivery_days"] = (
    orders_clean["order_delivered_customer_date"]
    - orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)
```

Estimated delivery time

```python
orders_clean["estimated_delivery_days"] = (
    orders_clean["order_estimated_delivery_date"]
    - orders_clean["order_purchase_timestamp"]
).dt.total_seconds() / (60 * 60 * 24)
```

Delivery delay

```python
orders_clean["delivery_delay_days"] = (
    orders_clean["order_delivered_customer_date"]
    - orders_clean["order_estimated_delivery_date"]
).dt.total_seconds() / (60 * 60 * 24)
```

```python
category_translation = pd.read_csv(
    DATA_PATH / "product_category_name_translation.csv"
)

category_translation.head()
```

```python
products_clean = products_clean.merge(
    category_translation,
    on="product_category_name",
    how="left"
)
```

```python
products_clean["product_category"] = (
    products_clean["product_category_name_english"]
    .fillna("Unknown")
)
```

```python
print("Orders:", orders_clean.shape)
print("Order Items:", order_items_clean.shape)
print("Products:", products_clean.shape)

print("\nNew order metrics:")
print(
    orders_clean[
        ["delivery_days", "estimated_delivery_days", "delivery_delay_days"]
    ].describe()
)

print("\nProduct categories:")
print(products_clean["product_category"].value_counts().head(10))
```

```python
# Check duplicate primary keys

print("Customer ID duplicates:",
      customers_clean["customer_id"].duplicated().sum())

print("Order ID duplicates:",
      orders_clean["order_id"].duplicated().sum())

print("Product ID duplicates:",
      products_clean["product_id"].duplicated().sum())

print("Seller ID duplicates:",
      sellers_clean["seller_id"].duplicated().sum())

print("Review ID duplicates:",
      reviews_clean["review_id"].duplicated().sum())
```
